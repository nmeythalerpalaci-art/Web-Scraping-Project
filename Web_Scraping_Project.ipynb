```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import csv



BIT_Faculty = 'https://bit.vt.edu/faculty/directory.html'


#BIT Faculty Server Status
BIT_response = requests.head(BIT_Faculty)
print(BIT_response)

#BIT Instructors URL
BIT_Faculty = 'https://bit.vt.edu/faculty/directory.html'
header1 = {'User-Agent':'Mozilla/5.0 (BIT 3474 Midterm)'}
response1 = requests.get(BIT_Faculty, headers=header1, timeout=10)

soup1 = BeautifulSoup(response1.text, 'html.parser')
links1 = soup1.find_all('a', class_='vt-list-item-title-link')

url1 = []
for link in links1:
  x = link.get('href')
  if x:
    url1.append(x)

print(url1)

BIT_Faculty = 'https://bit.vt.edu/faculty/directory.html'
header1 = {'User-Agent':'Mozilla/5.0 (BIT 3474 Midterm)'}
response1 = requests.get(BIT_Faculty, headers=header1, timeout=10)

BIT = BeautifulSoup(response1.text, 'html.parser')

#BIT Faculty Department
departments = BIT.find_all('span', class_='vt-logo-text')
dep1 = []
for department in departments:
  x = department.get_text(strip=True)
  if x:
    for i in range(61):
      dep1.append(x)
print(dep1)

#BIT College
college = BIT.find_all('a', class_='vt-breadcrumbs-link')
col1 = []
for i in college:
  x = i.get('title')
  if x:
    for i in range(61):
      col1.append(x)
print(col1)

#BIT intructor's first and last name
first_name = []
for name in url1:
  response = requests.get(name, headers=header1, timeout=10)
  soup = BeautifulSoup(response.text, 'html.parser')
  first = soup.find_all('h1', class_='vt-bio-name')
  for i in first:
    x = i.get_text(strip=True)
    if x:
      first_name.append(x)
  if not first:
    first_name.append('Not Available')

print(first_name)

#BIT intructor's title
title = []
for t in url1:
  response = requests.get(t,headers=header1,timeout=10)
  soup = BeautifulSoup(response.text, 'html.parser')
  a = soup.find_all('div', class_='vt-person-title')
  for i in a:
    x = i.get_text(strip=True)
    if x:
      title.append(x)
  if not a:
    title.append('Not Available')
print(title)

#BIT intructor's office building and number
office = []
for dep in url1:
  response = requests.get(dep, headers=header1, timeout=30)
  soup = BeautifulSoup(response.text, 'html.parser')
  building = soup.find_all('address', class_= 'vt-bio-contact-address')
  for i in building:
    x = i.get_text(',', strip=True)
    if x:
      office.append(x)
  if not building:
    office.append('Not Available')
print(office)

#BIT intructor's email
email = []
for e in url1:
  response = requests.get(e, headers=header1, timeout=30)
  soup = BeautifulSoup(response.text, 'html.parser')
  correo = soup.find_all('a', class_='vt-bio-contact-email')
  for i in correo:
    x = i.get_text('href', strip=True)
    if x:
      email.append(x)
  if not correo:
    email.append('Not Available')
print(email)

#BIT intructor's phone number
phone1 = []
for num in url1:
  response = requests.get(num, headers=header1, timeout=30)
  soup = BeautifulSoup(response.text, 'html.parser')
  number = soup.find_all('a', class_='vt-bio-phone-link')
  for i in number:
    x = i.get_text('href', strip=True)
    if x:
      phone1.append(x)
  if not number:
    phone1.append('Not Available')
print(phone1)


fieldnames = ['Department', 'College', 'First and Last Name',
              'Title', 'Office Building and Number', 'Email', 'Phone']


with open('BIT Department.csv', 'w', newline='') as csvfile:
  writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
  writer.writeheader()
  for d,c,f,t,o,e,p in zip(
      dep1, col1, first_name, title, office,email, phone1
  ):
      writer.writerow({
            'Department': d,
            'College': c,
            'First and Last Name': f,
            'Title': t,
            'Office Building and Number': o,
            'Email': e,
            'Phone': p
        })
```
